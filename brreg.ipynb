```python
import requests
import pandas as pd
from typing import List
```

```python
session = requests.Session()

def get_firms(session, industry_code: str, number_of_employees: str, size: int=10000, page:int=0, bankrupt:bool=False):
    """
    Get a list of firms in a specific industry code
    """
    base_url = "https://data.brreg.no/enhetsregisteret/api"
    endpoint = "/enheter"
    url = f"{base_url}{endpoint}"

    params = {
        "naeringskode": [industry_code],
        "fraAntallAnsatte": number_of_employees,
        "page": page,
        "size": size,
        "konkurs": bankrupt,
    }
    

    response = session.get(url, params=params)
    response.raise_for_status()
    return response.json()

def parse_firms(firms: List[dict]) -> pd.DataFrame:
    normalized_list = [pd.json_normalize(obj) for obj in firms]
    result_df = pd.concat(normalized_list, ignore_index=True)
    return result_df

df = parse_firms(get_firms(session, "43.210", "1").get("_embedded").get("enheter"))
df
```

```python
orgnrs = list(df["organisasjonsnummer"])
orgnrs
```

```python
def get_roles_for_firm(session, orgnr: str):
    """
    Get a list of roles for a specific firm
    """
    base_url = "https://data.brreg.no/enhetsregisteret/api"
    endpoint = f"/enheter/{orgnr}/roller"
    url = f"{base_url}{endpoint}"

    response = session.get(url)
    response.raise_for_status()
    return response.json()
```

```python
for orgnr in orgnrs:
    try:
        print(get_roles_for_firm(session, orgnr))
    except:
        print(f"Failed to get roles for {orgnr}")
```

```python
def get_fundamentals_for_firm(session, orgnr: str):
    """
    Get a list of fundamentals for a specific firm
    """
    base_url = "https://data.brreg.no/regnskapsregisteret"
    endpoint = f"/regnskap/{orgnr}"
    url = f"{base_url}{endpoint}"

    response = session.get(url)
    response.raise_for_status()
    return response.json()
```

```python
for orgnr in orgnrs:
    try:
        print(get_fundamentals_for_firm(session, orgnr))
    except:
        print(f"No data for {orgnr}")
```
